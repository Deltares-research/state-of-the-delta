## Define Areas

### Packages

```python
# Packages
from pathlib import Path

import geopandas as gpd
from resplotlib import rpc
```

### Define areas

```python
# Directories
repo_path = Path.cwd().parents[0]
dir_path_areas = repo_path / "data" / "01_areas"

# Area settings
name = "north_sea"
file_path_areas = dir_path_areas / f"{name}.geojson".lower().replace(" ", "_")
bounds = None
crs = "EPSG:3857"
```

### Show and draw areas

```python
# Read areas
if file_path_areas and file_path_areas.exists():
    print(f"{file_path_areas.name} already exist. Edit the areas if necessary.")
    gdf_areas = gpd.read_file(file_path_areas)
else:
    print(f"{file_path_areas.name} do not exist. Draw the areas if necessary.")
    gdf_areas = None

# Initialize map
m = rpc.Map(gdf=gdf_areas, bounds=bounds, draw_bounds=True, clear_on_draw=True)
m = rpc.explore_basemap(crs=crs, m=m, style="satellite")

# Show map
m
```

```python
# Get drawn geometries
gdf_areas = m.get_drawn_geometries(crs=crs)

# Save drawn geometries
gdf_areas.to_file(dir_path_areas / f"{name}.geojson".lower().replace(" ", "_"))
```
